```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

# Load CSV file
df = pd.read_csv("Figure_7B_TSC22D2_Phosphomutants_pI.csv")

# Set global font to Arial
plt.rcParams['font.family'] = 'Arial'

conditions = df.columns[1:]  # All condition columns

# --- Define fixed colors ---
final_colors = {cond: "#bababa" for cond in conditions}  # default: gray

final_colors["Isoelectric_point_pWT"] = "#6fd8c4"
final_colors["Isoelectric_point_40A"] = "#e8e8e8"
final_colors["Isoelectric_point_WT"] = "#595959"

# Plot setup
plt.figure(figsize=(3.7797, 2.9761))

for condition in conditions:
    plt.plot(df["pH"], df[condition],
             linestyle='-',
             linewidth=1,
             color=final_colors[condition],
             label=condition)

# Labels and formatting
plt.xlabel("pH", fontsize=12)
plt.ylabel("pI", fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.yticks(fontsize=10)

# Legend placed outside the plot
plt.legend().remove()
plt.tight_layout()

# Save & show the plot
plt.savefig("Figure_7B_TSC22D2_Phosphomutants_pI.pdf",
            format="pdf", dpi=300, bbox_inches="tight", transparent=True)
plt.show()
```
