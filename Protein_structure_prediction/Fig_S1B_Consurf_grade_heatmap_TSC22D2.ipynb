```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import seaborn as sns

# -----------------------------
# 1. Load the CSV file
# -----------------------------
file_path = "Figure_S1B_Consurf_grade_TSC22D2.csv"
df = pd.read_csv(file_path)

# Inspect columns to find residue position and grade
print("Columns:", df.columns.tolist())
print(df.head())

# Try to detect residue and grade column names automatically
residue_col = [c for c in df.columns if 'residue' in c.lower() or 'position' in c.lower()][0]
grade_col = [c for c in df.columns if 'grade' in c.lower() or 'consurf' in c.lower()][0]

# -----------------------------
# 2. Prepare data for plotting
# -----------------------------
grades = df[grade_col].astype(float)
positions = df[residue_col].astype(int)

# Reshape to 2D array (1 row, N columns) for heatmap
heatmap_data = grades.values.reshape(1, -1)

# -----------------------------
# 3. Define colormap (grey → teal)
# -----------------------------
cmap = mcolors.LinearSegmentedColormap.from_list(
    "white_teal", ["white", "#49c1bb"]
)

# -----------------------------
# 4. Plot the heatmap
# -----------------------------
plt.rcParams['font.family'] = 'Arial'
plt.figure(figsize=(7, 1.2))

sns.heatmap(
    heatmap_data,
    cmap=cmap,
    cbar_kws={"orientation": "horizontal", "label": "ConSurf Grade (0–9)"},
    xticklabels=False,
    yticklabels=False,
    linewidths=0
)

plt.xlabel("Residue Position", fontsize=10)
plt.title("TSC22D2 ConSurf Conservation Profile", fontsize=11, pad=10)
plt.tight_layout()

# Save high-resolution image (optional)
# plt.savefig("TSC22D2_ConSurf_heatmap.tiff", dpi=600, bbox_inches="tight", transparent=True)
plt.savefig("Figure_S1B_ConSurf_grade_TSC22D2.pdf", format="pdf", dpi=300, bbox_inches="tight", transparent=True)
plt.show()
```
