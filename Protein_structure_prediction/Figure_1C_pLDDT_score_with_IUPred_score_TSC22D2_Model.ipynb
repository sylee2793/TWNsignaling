```python
from pathlib import Path
import numpy as np
import json
import matplotlib.pyplot as plt
```

```python
# Step 1: Open the .defattr file
file_path = 'Figure_1C_TSC22D2_plddt.defattr'  # Replace with your actual file path

# Step 2: Read the file
with open(file_path, 'r') as file:
    lines = file.readlines()

# Step 3: Skip the first 3 lines
lines = lines[3:]  # This removes the top 3 rows

# Step 3: Save the remaining lines to a new file (or overwrite the original file)
with open('Figure_1C_TSC22D2_modified_plddt.defattr', 'w') as f:
    f.writelines(lines)
```

```python
# Replace with your actual file path
file_path = 'Figure_1C_TSC22D2_modified_plddt.defattr'

# Step 1: Open the file and read its contents line by line
residue_numbers = []
plddt_scores = []

with open(file_path, 'r') as f:
    lines = f.readlines()  # Read all lines into a list

    for line in lines:
        # Step 2: Split each line by whitespace (remove leading/trailing whitespace first)
        parts = line.strip().split()

        # Extract the residue number from the first part (e.g., /A:1@CA)
        residue_info = parts[0]
        residue_number = int(residue_info.split(':')[1].split('@')[0])  # Extract residue number

        # Extract the pLDDT score from the second part
        plddt_score = float(parts[1])

        # Append the results to respective lists
        residue_numbers.append(residue_number)
        plddt_scores.append(plddt_score)

# Step 3: Print the residue numbers and corresponding pLDDT scores
for residue, plddt in zip(residue_numbers, plddt_scores):
    print(f"Residue {residue}: pLDDT {plddt}")
```

```python
# Step 1: Load your .json file
with open('E:/github/Protein structure prediction/TSC22D2/TSC22D2_IUPred3_Score.json', 'r') as f:
    data = json.load(f)

# Display the top-level keys to understand the structure
print(data.keys())
```

```python
# Check data types of the elements
for key, value in data.items():
    print(f"{key}: {type(value)}")

# Optional: Display a small sample of each key's content
for key, value in data.items():
    if isinstance(value, list):
        print(f"{key}: {value[:5]}")  # Display the first 5 elements
    elif isinstance(value, dict):
        print(f"{key}: {list(value.keys())[:5]}")  # Display the first 5 keys of a dictionary
    else:
        print(f"{key}: {value}")  # Display value if it's not a list or dict
```

```python
iupred_score = data.get('iupred2')
print(iupred_score)
```

```python
# Create a color array based on conditions for each residue
colors = []

# Define the coloring rules for different ranges of pLDDT scores
for motif in residue_numbers:
    if  7 <= motif <= 19:  # TbrW1 motif
        colors.append('blue')
    elif 160 <= motif <= 163:  #  motif
        colors.append('green')
    elif 172 <= motif <= 190:  # TbrN motif
        colors.append('red')
    elif 694 <= motif <= 749:  # TSC22D box
        colors.append('orange')
    else:  # other residues
        colors.append('white')
```

```python
import matplotlib.pyplot as plt

# Create the figure and first axis
fig, ax = plt.subplots(figsize=(6.8801, 0.9973))

# Set x-axis range
x_min = 0
x_max = 780
xtick_interval = 100
plt.xlim(x_min, x_max)
plt.xticks(np.arange(x_min, x_max+1, xtick_interval), rotation=45)

# Plot data for the first y-axis
ax.scatter(residue_numbers, plddt_scores, c=colors, s=10, marker='o', edgecolor='black')
ax.set_xlabel('Residue', fontsize=14, family='Arial', labelpad=5)
ax.set_ylabel('pLDDT (Cα)', fontsize=14, family='Arial', labelpad=5)
ax.tick_params(axis='both', labelsize=12)

# Create the second y-axis using twinx
ax2 = ax.twinx()

# Plot data for the second y-axis
ax2.plot(residue_numbers, iupred_score, c='gray')
ax2.set_ylabel('IUPred score', fontsize=14, family='Arial', labelpad=20, rotation=-90)
ax2.tick_params(axis='y', labelsize=12)

# Set y-limits for both axes
ax.set_ylim(0, 100)
ax2.set_ylim(0, 1)
ax.spines['bottom'].set_bounds(1, 780)

# Customize the plot
plt.title('pLDDT confidence score with IUPred disorder score', fontsize=16, family='Arial')

# Save the plot as a PDF
plt.savefig('Figure_1C_TSC22D2_pLDDT_Ca_with_IUPred_Score_Model_0.pdf', format='pdf', bbox_inches='tight')

# Show the plot
plt.show()
```
