```python
import pandas as pd

df = pd.read_csv('TSC22D1_AF-Q15714-F1-aa-substitutions.csv')
df['Residue'] = df['protein_variant'].str.extract(r"[A-Za-z](\d+)[A-Za-z]")[0].astype(int)
df['WT']      = df['protein_variant'].str[0]
df['AA']      = df['protein_variant'].str[-1]
df['Pathogenicity'] = df.apply(
    lambda row: 'Reference' if row['WT'] == row['AA'] else row['am_pathogenicity'],
    axis=1
)
df_formatted = df[['Residue', 'WT', 'AA', 'Pathogenicity']]
df_formatted.to_csv('TSC22D1_AF-Q15714-F1-formatted.csv', index=False)

print(df_formatted.head())
```
