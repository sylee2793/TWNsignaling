```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mcolors
from matplotlib.colors import Normalize
import matplotlib.cm as cm
```

```python
# Load the data from CSV file
data = pd.read_csv('Figure_1B_TSC22D2_AP-MS.csv')

# Define the categories and the three sets of y-values
proteins = data['Protein']
isotonic = data['Isotonic']
sorbitol = data['Sorbitol']
sorbitol_wnk463 = data['Sorbitol_WNK463']

# Define the highlighting columns
isotonic_p = data['Isotonic_P']
sorbitol_p = data['Sorbitol_P']
sorbitol_wnk463_p = data['Sorbitol_WNK463_P']

# Set up colormap and normalization for gradient
cmap = mcolors.LinearSegmentedColormap.from_list(
    "white_to_gold",
    ["#ffffff", "#efd93a"]
)
norm = mcolors.Normalize(vmin=min(isotonic_p.min(), sorbitol_p.min(), sorbitol_wnk463_p.min()), 
                         vmax=max(isotonic_p.max(), sorbitol_p.max(), sorbitol_wnk463_p.max()))

# Define colors for lines for each condition
line_colors = ['#e8e8e8', '#49c1bb', '#b5b5b5']

# Plot setup
fig, ax = plt.subplots(figsize=(5.212, 3.3069))

# Define small offsets to separate the dots within each protein category
offsets = np.linspace(-0.3, 0.3, 3)

# Loop through each condition and plot the lollipop lines and dots with gradient color
for i, (y, p_values, line_color) in enumerate(zip(
        [isotonic, sorbitol, sorbitol_wnk463], 
        [isotonic_p, sorbitol_p, sorbitol_wnk463_p],
        line_colors)):
    
    # Map P-values to colors using the colormap and normalization for gradient dots
    colors = cmap(norm(p_values))
    
    # Plot the lollipop line with specific color and dot with gradient color
    ax.vlines(x=np.arange(len(proteins)) + offsets[i], ymin=0, ymax=y, color=line_color, linewidth=2, zorder=1)
    ax.scatter(np.arange(len(proteins)) + offsets[i], y, color=colors, s=15, edgecolors='k', linewidth=0.5, zorder=3, label=['Isotonic', 'Sorbitol', 'Sorbitol_WNK463'][i])

# Set y-axis range
yticks = np.arange(0, 101, 20)
plt.yticks(yticks)

# Separate the (0,0) coordinate by moving the spines
ax = plt.gca()
ax.spines['left'].set_position(('outward', 5))  # Move the y-axis spine outward
ax.spines['bottom'].set_position(('outward', 2))  # Move the x-axis spine outward
ax.spines['left'].set_bounds(0, 100)

# Hide the top and right spines
ax.spines['top'].set_color('none')
ax.spines['right'].set_color('none')

# Customizing the plot
ax.set_xticks(np.arange(len(proteins)))
ax.set_xlim(-0.5, len(proteins) - 0.5)  # Adjust starting and ending x-limits for a tighter fit
ax.set_xticklabels(proteins, rotation=90)
ax.set_ylabel("Total spectral count", fontsize=14, family='Arial')
ax.set_title("TSC22D2 AP-MS", fontsize=16, family='Arial')

# Use tight layout to reduce whitespace
plt.tight_layout()

# Save the plot as a PDF
plt.savefig('Figure_1B_TSC22D2_AP-MS_PTMsearch_Lollipop.pdf', format='pdf', bbox_inches='tight')

# Show the main plot

plt.show()

# Create a separate legend plot
fig_legend, ax_legend = plt.subplots(figsize=(2, 1))
ax_legend.axis('off')  # Turn off axis for the legend-only plot


# Add legend entries
for color, label in zip(line_colors, ['Isotonic', 'Sorbitol', 'Sorbitol+WNK463']):
    ax_legend.plot([], [], color=color, linestyle='-', linewidth=2, label=label)

# Customize the legend with specific font size and style
ax_legend.legend(
    loc='center', 
    frameon=False, 
    fontsize=12,  # Set the font size here
    title="Conditions",
    title_fontsize=12,  # Font size for the title
    prop={'family': 'Arial'}  # Font family, style, and weight
)

# Save the plot as a PDF
plt.savefig('Figure_1B_TSC22D2_AP-MS_PTMsearch_Lollipop_Legend1.pdf', format='pdf', bbox_inches='tight')

# Set up the colormap and normalization
cmap = mcolors.LinearSegmentedColormap.from_list(
    "white_to_gold",
    ["#ffffff", "#efd93a"]
)
norm = Normalize(vmin=0, vmax=15)  # Adjust min and max values as needed

# Create a circular gradient color wheel
fig, ax = plt.subplots(figsize=(2, 2), subplot_kw=dict(polar=True))
theta = np.linspace(0, 2 * np.pi, 100)

# Generate radial values from 0 to 1 (for radius) and use it as the color intensity
r = np.linspace(0.2, 1, 100)  # Range of radius values
T, R = np.meshgrid(theta, r)

# Map radial values to the gradient colors based on the P-value range
Z = R * 15  # Scale radius to the P-value range for color mapping
colors = cmap(norm(Z))

# Plot the gradient as a filled contour plot
ax.pcolormesh(T, R, Z, cmap=cmap, norm=norm, shading='auto')

# Customize the circle appearance
ax.set_yticklabels([])  # Remove radial axis labels
ax.set_xticklabels([])  # Remove angular axis labels
ax.spines['polar'].set_visible(False)  # Hide the polar spine
ax.set_title("Phospho-peptide \n spectral count", fontsize=12)

# Save the plot as a PDF
plt.savefig('Figure_1B_TSC22D2_AP-MS_PTMsearch_Lollipop_Legend2.pdf', format='pdf', bbox_inches='tight')

plt.show()
```
