```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Load data
file_path = "Figure_S2I_Kidney_WT_vs_TSC22D-HOM.csv"
df = pd.read_csv(file_path)

# Reshape to long format
df_long = df.melt(var_name='Condition', value_name='Foci Count')

# Define genotype (WT / TSC22D-HOM)
df_long['Genotype'] = df_long['Condition'].apply(lambda x: 'WT' if 'WT' in x else 'TSC22D-HOM')

# Colors for violins
violin_colors = {"WT": "#bababa", "TSC22D-HOM": "#49c1bb"}
condition_order = df.columns.tolist()  # Maintain original order

plt.rcParams['font.family'] = 'Arial'
fig, ax = plt.subplots(figsize=(3, 3))

# --- Violin plot (full first) ---
vp = sns.violinplot(
    x='Condition', y='Foci Count', data=df_long,
    hue='Genotype', order=condition_order,
    inner=None, cut=0, scale='width', palette=violin_colors,
    linewidth=0, dodge=False, ax=ax
)

# --- Clip violins to left half only (half-violin style) ---
for violin in ax.collections:
    path = violin.get_paths()[0]
    vertices = path.vertices
    center_x = np.median(vertices[:, 0])  # Find center x
    vertices[:, 0] = np.minimum(vertices[:, 0], center_x)  # Keep only left half
    path.vertices = vertices

# --- Boxplot (white fill) ---
sns.boxplot(
    x='Condition', y='Foci Count', hue='Genotype', data=df_long,
    order=condition_order, showcaps=False, showfliers=False, width=0.35,
    boxprops={'facecolor': 'white', 'edgecolor': 'black', 'zorder': 4},
    medianprops={'color': 'black', 'linewidth': 1.2},
    whiskerprops={'color': 'black'}, capprops={'color': 'black'},
    palette=['white'], dodge=False, ax=ax
)

# --- Stripplot (small dots, black edge) ---
sns.stripplot(
    x='Condition', y='Foci Count', hue='Genotype', data=df_long,
    order=condition_order, jitter=True, size=0.5, alpha=1,
    palette=['white'], edgecolor='black', linewidth=0.6,
    dodge=False, ax=ax
)

# --- Mean line per condition ---
group_means = df_long.groupby('Condition')['Foci Count'].mean()
for i, cond in enumerate(condition_order):
    ax.hlines(
        y=group_means[cond], xmin=i - 0.2, xmax=i + 0.2,
        colors='black', linewidth=1.2, zorder=6
    )

# Remove legends
if ax.get_legend():
    ax.get_legend().remove()

# --- Vertical dashed lines to separate compartments ---
# Assuming there are 3 compartments (Cell, Nuclei, Cytoplasm)
for pos in [1.5, 3.5]:
    ax.axvline(x=pos, color='black', linestyle='--', linewidth=1)

# --- Labels ---
plt.xlabel("Condition")
plt.ylabel("Area (µm\u00b2) per cell")
plt.xticks(rotation=45, ha='right')

# --- Save figure ---
plt.savefig("Figure_S2I_Kidney_Measurements.pdf",
            format="pdf", dpi=300, bbox_inches="tight", transparent=True)
plt.show()
```

```python
import pandas as pd
from scipy import stats

# --- Load data ---
file_path = "Figure_2I_Kidney_WT_vs_TSC22D-HOM.csv"
df = pd.read_csv(file_path)

# --- Define paired columns ---
comparisons = {
    "Cell": ("WT_Kidney_Cell", "TSC22D2_Kidney_Cell"),
    "Nuclei": ("WT_Kidney_Nuclei", "TSC22D2_Kidney_Nuclei"),
    "Cytoplasm": ("WT_Kidney_Cytoplasm", "TSC22D2_Kidney_Cytoplasm")
}

# --- Perform Welch's t-test for each compartment ---
results = []
for comp, (wt_col, hom_col) in comparisons.items():
    wt_vals = df[wt_col].dropna()
    hom_vals = df[hom_col].dropna()

    t_stat, p_val = stats.ttest_ind(wt_vals, hom_vals, equal_var=False)  # Welch's t-test
    df_wt, df_hom = len(wt_vals) - 1, len(hom_vals) - 1
    mean_wt, mean_hom = wt_vals.mean(), hom_vals.mean()
    sd_wt, sd_hom = wt_vals.std(), hom_vals.std()

    results.append({
        "Compartment": comp,
        "WT_mean": mean_wt,
        "WT_SD": sd_wt,
        "TSC22D2-HOM_mean": mean_hom,
        "TSC22D2-HOM_SD": sd_hom,
        "t_statistic": t_stat,
        "p_value": p_val,
        "n_WT": len(wt_vals),
        "n_HOM": len(hom_vals)
    })

# --- Convert to DataFrame and save ---
results_df = pd.DataFrame(results)
print(results_df)
```
