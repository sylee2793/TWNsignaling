```python
import pandas as pd

# Load csv file from lacal repository
df1 = pd.read_csv('Figure_S6C_WNK463_CrowdFRET_Processed.csv')
print(df1.head())  # Display the first few rows of the DataFrame
```

```python
# install matplotlib library
! pip install matplotlib
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Apply rolling mean for smoothing
df1['Isotonic_Mean_smooth'] = df1['Isotonic_Mean'].rolling(window=3, center=True).mean()
df1['WNK463_1uM_Mean_smooth'] = df1['WNK463_1uM_Mean'].rolling(window=3, center=True).mean()
df1['WNK463_2uM_Mean_smooth'] = df1['WNK463_2uM_Mean'].rolling(window=3, center=True).mean()
df1['WNK463_5uM_Mean_smooth'] = df1['WNK463_5uM_Mean'].rolling(window=3, center=True).mean()
df1['Isotonic_SD_smooth'] = df1['Isotonic_SD'].rolling(window=3, center=True).mean()
df1['WNK463_1uM_SD_smooth'] = df1['WNK463_1uM_SD'].rolling(window=3, center=True).mean()
df1['WNK463_2uM_SD_smooth'] = df1['WNK463_2uM_SD'].rolling(window=3, center=True).mean()
df1['WNK463_5uM_SD_smooth'] = df1['WNK463_5uM_SD'].rolling(window=3, center=True).mean()

# Create the plot
plt.figure(figsize=(5,3))
plt.plot(df1['TIME'], df1['Isotonic_Mean_smooth'], label='Isotonic', color='#c0c0c0')
plt.fill_between(df1['TIME'], df1['Isotonic_Mean_smooth'] - df1['Isotonic_SD_smooth'], df1['Isotonic_Mean_smooth'] + df1['Isotonic_SD_smooth'], color='#c0c0c0', alpha=0.2)
plt.plot(df1['TIME'], df1['WNK463_1uM_Mean_smooth'], label='WNK463 1 μM', color='#aec7e8')
plt.fill_between(df1['TIME'], df1['WNK463_1uM_Mean_smooth'] - df1['WNK463_1uM_SD_smooth'], df1['WNK463_1uM_Mean_smooth'] + df1['WNK463_1uM_SD_smooth'], color='#aec7e8', alpha=0.2)
plt.plot(df1['TIME'], df1['WNK463_2uM_Mean_smooth'], label='WNK463 2 μM', color='#5178B3')
plt.fill_between(df1['TIME'], df1['WNK463_2uM_Mean_smooth'] - df1['WNK463_2uM_SD_smooth'], df1['WNK463_2uM_Mean_smooth'] + df1['WNK463_2uM_SD_smooth'], color='#5178B3', alpha=0.2)
plt.plot(df1['TIME'], df1['WNK463_5uM_Mean_smooth'], label='WNK463 5 μM', color='#1f77b4')
plt.fill_between(df1['TIME'], df1['WNK463_5uM_Mean_smooth'] - df1['WNK463_5uM_SD_smooth'], df1['WNK463_5uM_Mean_smooth'] + df1['WNK463_5uM_SD_smooth'], color='#1f77b4', alpha=0.2)

# Set x range
xticks = np.arange(0, 1250, 200)
plt.xticks(xticks)
plt.tick_params(axis='both', labelsize=12)

# Set y-axis range
yticks = np.arange(0.4, 1.9, 0.2)  # Start at 0.5, end at 1.6, with interval of 0.2. Include 1.5 by extending the range slightly
plt.yticks(yticks)

# Separate the (0,0) coordinate by moving the spines
ax = plt.gca()
ax.spines['left'].set_position(('outward', 5))  # Move the y-axis spine outward
ax.spines['bottom'].set_position(('outward', 20))  # Move the x-axis spine outward

# Hide the top and right spines
ax.spines['top'].set_color('none')
ax.spines['right'].set_color('none')

# Set the bounds for the bottom and left spines to start from the first tick
ax.spines['bottom'].set_bounds(0, 1250)
ax.spines['left'].set_bounds(0.4, 1.8)

# Customize the plot
plt.title('Crowd FRET', fontsize=16, family='Arial')
plt.xlabel('Time', fontsize=14, family='Arial', labelpad=5)
plt.ylabel('Normalized mCitrine/mCerulean' + '\n' + '(525/475 nm)', fontsize=14, family='Arial', labelpad=5)
plt.grid(False)

# Customize the legend
plt.legend(
    frameon=False,
    loc='center left',
    bbox_to_anchor=(1,0.5),
    prop={'family': 'Arial', 'size': 12}
)

#Adjust the area of the legend
plt.subplots_adjust(right=0.75)  # Adjust the right boundary to make space

# Save the plot as a PDF
plt.savefig('Figure_S6C_CrowdFRET_WNK463.pdf', format='pdf', bbox_inches='tight')

# Display the plot
plt.show()
```
