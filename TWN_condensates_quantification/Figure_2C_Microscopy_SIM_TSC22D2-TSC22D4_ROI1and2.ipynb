```python
from pathlib import Path
import numpy as np
import json
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load data
data = pd.read_csv('Figure_2C_SIM_TSC22D2-mScarlet_mNeonGreen-TSC22D4_ROI_1.csv')

# Display the first few rows to check the data
data.head()
```

```python
# Create figure and axes
fig, ax = plt.subplots(figsize=(2.44, 1.72), sharex=True)

# Define labels and colors for each dataset
labels = ['TSC22D2', 'TSC22D4']  # replace with your actual column names
colors = ['#d80c0d', '#1c6f31']

# Plot both datasets in the same axis
for i, column in enumerate(labels):
    ax.plot(
        data['Position_um'],
        data[column],
        'o-',
        color=colors[i],
        alpha=0.4,
        label=column  # Add legend label
    )

# Set axis labels
ax.set_xlabel('Distance (μm)')
ax.set_ylabel('Value')  # Customize as needed

# Set y-limits to include all data
y_min = data[labels].min().min()
y_max = data[labels].max().max()
ax.set_ylim(y_min - 5, y_max + 5)

# Show legend
ax.legend()

# Adjust layout
plt.tight_layout()

# Save figure
plt.savefig('Figure_2C_SIM_TSC22D2-TSC22D4_ROI_1_combined.pdf', format='pdf', bbox_inches='tight')

plt.show()
```
