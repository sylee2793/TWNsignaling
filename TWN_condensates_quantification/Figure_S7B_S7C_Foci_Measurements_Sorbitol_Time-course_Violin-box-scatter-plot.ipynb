```python
# ================================================================
# Extended Data Figure 7b: Foci number time-course analysis
# (Preserving original half-violin + box + strip style)
# ================================================================
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from scipy import stats

# ------------------------------------------------
# 1. Load data
# ------------------------------------------------
file_path = "Figure_S7B_Foci_number_Time-course.csv"
df = pd.read_csv(file_path)

# Automatically reshape to long format if needed
# Assumes each column = time point, each row = replicate
df_long = df.melt(var_name='Time', value_name='Foci Count')

# Sort time numerically if possible
try:
    df_long['Time'] = df_long['Time'].astype(float)
    df_long = df_long.sort_values('Time')
except:
    pass

time_order = df_long['Time'].unique().tolist()

plt.rcParams['font.family'] = 'Arial'
fig, ax = plt.subplots(figsize=(4, 3.2))

# --- Violin plot (full first) ---
vp = sns.violinplot(
    x='Time', y='Foci Count', data=df_long,
    inner=None, cut=0, scale='width', palette=["#e8e8e8"],
    linewidth=0, dodge=False, ax=ax
)

# --- Clip violins to left half only ---
for violin in ax.collections:
    try:
        path = violin.get_paths()[0]
        vertices = path.vertices
        center_x = np.median(vertices[:, 0])
        vertices[:, 0] = np.minimum(vertices[:, 0], center_x)
        path.vertices = vertices
    except Exception:
        pass

# --- Boxplot (white fill) ---
sns.boxplot(
    x='Time', y='Foci Count', data=df_long,
    showcaps=False, showfliers=False, width=0.35,
    boxprops={'facecolor': 'white', 'edgecolor': 'black', 'zorder': 4},
    medianprops={'color': 'black', 'linewidth': 1.2},
    whiskerprops={'color': 'black'}, capprops={'color': 'black'},
    palette=['white'], dodge=False, ax=ax
)

# --- Stripplot (white dots with black edge) ---
sns.stripplot(
    x='Time', y='Foci Count', data=df_long,
    jitter=True, size=3.5, alpha=1,
    palette=['white'], edgecolor='black', linewidth=0.6,
    dodge=False, ax=ax
)

# --- Mean as horizontal line ---
group_means = df_long.groupby('Time')['Foci Count'].mean()
for i, cond in enumerate(time_order):
    ax.hlines(
        y=group_means[cond], xmin=i - 0.2, xmax=i + 0.2,
        colors='black', linewidth=1.2, zorder=6
    )

# Labels and cleanup
plt.xlabel("Time (min)")
plt.ylabel("Total number of foci / cell")
plt.xticks(rotation=45, ha='right')
if ax.get_legend():
    ax.get_legend().remove()

# Save
plt.savefig("Figure_S7B_Foci_number_Time-course.pdf",
            format="pdf", dpi=300, bbox_inches="tight", transparent=True)
plt.show()
```
