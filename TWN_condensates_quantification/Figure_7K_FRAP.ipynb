```python
import pandas as pd
import matplotlib.pyplot as plt

# =====================
# Load CSV
# =====================
file_path = "Fig_7K_FRAP.csv"
df = pd.read_csv(file_path)

# =====================
# Colors
# =====================
color_wt = "#6fd8c4"
color_40a = "#e8e8e8"

# =====================
# Global font setting
# =====================
plt.rcParams["font.family"] = "Arial"

# =====================
# Plot
# =====================
fig, ax = plt.subplots(figsize=(7.5, 4))

# --- WT ---
ax.scatter(
    df["Frame"], df["WT_Y1"],
    color=color_wt, alpha=0.5, s=30,
    label="WT Y1"
)
ax.plot(
    df["Frame"], df["WT_Y2"],
    color=color_wt, linewidth=3,
    label="WT Y2"
)

# --- 40A ---
ax.scatter(
    df["Frame"], df["40A_Y1"],
    color=color_40a, alpha=0.5, s=30,
    label="40A Y1"
)
ax.plot(
    df["Frame"], df["40A_Y2"],
    color=color_40a, linewidth=3,
    label="40A Y2"
)

# =====================
# Formatting (force Arial everywhere)
# =====================
ax.set_xlabel("Time (sec)", fontname="Arial")
ax.set_ylabel("Value", fontname="Arial")

ax.tick_params(axis="both", labelsize=9)
for label in ax.get_xticklabels() + ax.get_yticklabels():
    label.set_fontname("Arial")

legend = ax.legend(frameon=False, fontsize=9)
for text in legend.get_texts():
    text.set_fontname("Arial")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.margins(x=0)
plt.tight_layout()
plt.savefig("Figure_7K_FRAP_WT_vs_40A.pdf", format="pdf", dpi=300, bbox_inches="tight", transparent=True)
plt.show()
```
